```python

import matplotlib.pyplot as plt

# Filter out 'UNKNOWN' weapons (case-insensitive)
filtered_data = read_crime_data[read_crime_data['Weapon Desc'].str.upper() != 'UNKNOWN']

# Count the frequency of each weapon
weapon_counts = filtered_data['Weapon Desc'].value_counts()

# Calculate percentages
weapon_percentages = (weapon_counts / weapon_counts.sum()) * 100

# Plot the top 10 weapons by percentage
plt.figure(figsize=(12, 8))
weapon_percentages.head(10).plot(kind='bar', color='mediumseagreen')

plt.title('Top 10 Weapons Used in Crime Data (Excluding Unknown)')
plt.xlabel('Weapon Description')
plt.ylabel('Percentage of Incidents (%)')
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels for better visibility
plt.tight_layout()
plt.show()
```

```python
# Create a new column to indicate weapon involvement
read_crime_data['Weapon Involved'] = read_crime_data['Weapon Desc'].str.upper() != 'UNKNOWN'

# Group by 'Crm Cd Desc' (crime description) and 'Weapon Involved'
weapon_stats = read_crime_data.groupby(['Crm Cd Desc', 'Weapon Involved']).size().unstack(fill_value=0)

# Calculate the percentage of incidents involving a weapon for each crime type
weapon_stats['Weapon Involved %'] = (weapon_stats[True] / weapon_stats.sum(axis=1)) * 100

# Sort by highest percentage of weapon involvement
weapon_stats_sorted = weapon_stats.sort_values(by='Weapon Involved %', ascending=False)

# Plot
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 8))
weapon_stats_sorted['Weapon Involved %'].head(10).plot(kind='bar', color='orchid')  # Top 15 for better visibility
plt.title('Percentage of Incidents Involving Weapons by Crime Type')
plt.xlabel('Crime Type')
plt.ylabel('Percentage of Incidents (%)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Sort for weapon involvement
top_weapon_involved = weapon_stats_sorted['Weapon Involved %'].head(10)

# Sort for *least* weapon involvement (i.e., top 10 crimes without weapons)
least_weapon_involved = weapon_stats_sorted['Weapon Involved %'].tail(10)

# Plotting
fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(14, 12))

# Top 10 crimes with weapon
top_weapon_involved.plot(kind='bar', color='firebrick', ax=axes[0])
axes[0].set_title('Top 10 Crime Types WITH Weapon Involvement')
axes[0].set_ylabel('Weapon Involvement (%)')
axes[0].set_xticklabels(top_weapon_involved.index, rotation=45, ha='right')

# Top 10 crimes without weapon
least_weapon_involved.plot(kind='bar', color='seagreen', ax=axes[1])
axes[1].set_title('Top 10 Crime Types WITHOUT Weapon Involvement')
axes[1].set_ylabel('Weapon Involvement (%)')
axes[1].set_xticklabels(least_weapon_involved.index, rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming your DataFrame is already loaded as `read_crime_data`

# Step 1: Create a column to indicate weapon involvement
read_crime_data['Weapon Involved'] = read_crime_data['Weapon Desc'].str.upper() != 'UNKNOWN'

# Step 2: Group by Crime Description and Weapon Involvement
weapon_stats = read_crime_data.groupby(['Crm Cd Desc', 'Weapon Involved']).size().unstack(fill_value=0)

# Step 3: Calculate percentages
weapon_stats['Total'] = weapon_stats.sum(axis=1)
weapon_stats['Weapon Involved %'] = (weapon_stats[True] / weapon_stats['Total']) * 100
weapon_stats['No Weapon %'] = (weapon_stats[False] / weapon_stats['Total']) * 100

# Step 4: Sort
weapon_stats_sorted = weapon_stats.sort_values(by='Weapon Involved %', ascending=False)
non_weapon_sorted = weapon_stats.sort_values(by='No Weapon %', ascending=False)

# Step 5: Extract top 10 from both
top_with_weapon = weapon_stats_sorted['Weapon Involved %'].head(10)
top_without_weapon = non_weapon_sorted['No Weapon %'].head(10)

# Step 6: Plot
fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(14, 12))

# Plot top 10 with weapon
top_with_weapon.plot(kind='bar', color='crimson', ax=axes[0])
axes[0].set_title('Top 10 Crime Types Involving Weapons')
axes[0].set_ylabel('Weapon Involved (%)')
axes[0].set_xticklabels(top_with_weapon.index, rotation=45, ha='right')

# Plot top 10 without weapon
top_without_weapon.plot(kind='bar', color='darkgreen', ax=axes[1])
axes[1].set_title('Top 10 Crime Types NOT Involving Weapons')
axes[1].set_ylabel('No Weapon Involved (%)')
axes[1].set_xticklabels(top_without_weapon.index, rotation=45, ha='right')

plt.tight_layout()
plt.show()
```
