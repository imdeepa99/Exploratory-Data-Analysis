```python
import pandas as pd
pd.set_option('display.max_columns', None)
read_crime_data=pd.read_csv("../csv_files/cleaned_crime_data_2020_now.csv")
read_crime_data.head(8)
```

```python
# Data visualization

# Count the frequency of each crime type
crime_counts = read_crime_data['Crm Cd Desc'].value_counts()
crime_counts
```

```python
import matplotlib.pyplot as plt
```

```python
crime_counts.head(10).plot(kind='pie', autopct='%1.1f%%', figsize=(8, 8), startangle=90)
plt.title('Top 10 Most Frequent Crime Types')
plt.ylabel('')  # Hide the label on y-axis
plt.show()
```

```python
weapon_count = read_crime_data.groupby('Weapon Desc').size()
print(weapon_count)
```

```python
import matplotlib.pyplot as plt

# Count the frequency of each weapon description (value_counts ignores NaNs by default)
weapon_counts = read_crime_data['Weapon Desc'].value_counts()

# Sort the values in descending order and plot the top 10
plt.figure(figsize=(12, 8))
weapon_counts.sort_values(ascending=False).head(10).plot(kind='bar', color='skyblue')

plt.title('Top 10 Weapons Used in Crime Data (including unknown)')
plt.xlabel('Weapon Description')
plt.ylabel('Number of Incidents')
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels for better visibility
plt.tight_layout()
plt.show()
```

```python
# Remove rows where 'Weapon Desc' is 'UNKNOWN' (case-insensitive, just in case)
weapon_counts = read_crime_data[read_crime_data['Weapon Desc'].str.upper() != 'UNKNOWN']['Weapon Desc'].value_counts()

# Sort and plot the top 10
plt.figure(figsize=(12, 8))
weapon_counts.head(10).plot(kind='bar', color='lightcoral')

plt.title('Top 10 Weapons Used in Crime Data (Excluding Unknown)')
plt.xlabel('Weapon Description')
plt.ylabel('Number of Incidents')
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels for better visibility
plt.tight_layout()
plt.show()
```
