```python
import pandas as pd
pd.set_option('display.max_columns',None)
read_crime_data=pd.read_csv("../csv_files/raw_crime_data_2020_now.csv")
cleaned_data=read_crime_data.copy()
cleaned_data.head()
```

```python

print(type(cleaned_data))
cleaned_data.columns
cleaned_data.head()
```

```python
cleaned_data.columns
```

```python
# finding the null data
null_data=cleaned_data.isnull().sum()
null_data
```

```python
null_percentage=(null_data/len(cleaned_data))*100
null_percentage
```

```python
# Handling the null data

# for 0-5% missing values: simply drop the rows

cleaned_data.dropna(
    subset=['Premis Cd', 'Premis Desc', 'Crm Cd 1'],
    how='any',
    inplace=True
)
```

```python
cleaned_data.fillna({'Mocodes':'Unknown'},inplace=True)
cleaned_data.fillna({'Weapon Desc':'Unknown'},inplace=True)
cleaned_data.fillna({'Weapon Used Cd':'Unknown'},inplace=True)
```

```python
cleaned_data.replace({'Vict Sex': {'X': 'Unknown'}}, inplace=True)

cleaned_data.fillna({'Vict Sex':'Unknown'},inplace=True)

cleaned_data.fillna({'Vict Descent':'Unknown'},inplace=True)
```

```python

 #since the crmcd2, crmcd3 and crmcd 4 and  cross street has the 84% above missing values.. the analysis might ne
#bias so its safe to drop the entire column
#Weapon Used Cd  vs Weapon Desc
# also i have to drop the Weapon Used Cd column cause it has null values plus Weapon Desc column provides all the necessary information about the weapon
#If Weapon Desc already contains enough information (i.e., the description of the weapon), the Weapon Used Cd becomes redundant
# if Weapon Desc says "Gun", you don’t really need the numeric code 400 in the Weapon Used Cd
#text-based descriptions (like "Gun", "Knife", etc.) is easier for human understanding


cleaned_data.drop(["Crm Cd 2","Crm Cd 3","Crm Cd 4","Weapon Used Cd" and "Cross Street"],axis=1,inplace=True)
```
